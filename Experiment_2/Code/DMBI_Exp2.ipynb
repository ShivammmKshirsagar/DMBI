```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Force pandas to display all rows without truncating
pd.set_option('display.max_rows', None)

# Load the raw dataset
df = pd.read_csv('rapido_rides_raw.csv')

# Display the entire unprocessed dataset
df
```

```python
# Check for missing values before cleaning
print("Missing Values Before Cleaning:")
print(df.isnull().sum())
```

```python
# Fill missing values in numerical columns with the column mean
df['Ride_Distance_km'] = df['Ride_Distance_km'].fillna(df['Ride_Distance_km'].mean())
df['Total_Fare'] = df['Total_Fare'].fillna(df['Total_Fare'].mean())

# Verify missing values are handled
print("Missing Values After Filling:")
print(df.isnull().sum())

# Show the entire table
df
```

```python
print(f"Row count before removing duplicates: {len(df)}")

# Remove duplicate rows
df = df.drop_duplicates()

print(f"Row count after removing duplicates: {len(df)}")

# Show the entire table
df
```

```python
# Convert text categories into numerical values
label_encoder = LabelEncoder()
df['Vehicle_Type'] = label_encoder.fit_transform(df['Vehicle_Type'])
df['Payment_Type'] = label_encoder.fit_transform(df['Payment_Type'])

# Show the relevant columns for all rows
df[['Ride_ID', 'Vehicle_Type', 'Payment_Type']]
```

```python
# Standardize numerical features to bring them to a similar scale
scaler = StandardScaler()
df[['Ride_Distance_km', 'Total_Fare']] = scaler.fit_transform(df[['Ride_Distance_km', 'Total_Fare']])

# Display the final, fully preprocessed dataset
df
```
